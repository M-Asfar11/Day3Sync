```python
# 3. Challenge 3: The "Difficult" One – String & Data Cleaning
# -Goal: Handling real-world "dirty" data in Pandas.
# -The Setup: Give them this messy DataFrame:

# # df = pd.DataFrame({
# #     'Date': ['2023-01-01', '2023-01-02', 'Missing', '2023-01-04'],
# #     'Revenue': ['$1,200', '$2,500', 'ERROR', '$1,800'],
# #     'Active_Users': [150, np.nan, 200, 175]
# # })    

                # -The Mission: Write a script to "sanitize" this data:
                # -The Date Trap: Convert the Date column to datetime objects. (They will need to handle the "Missing" string—hint: errors='coerce').
                # -The Currency Trap: Convert Revenue into a clean float. They must remove the $ and the , and handle the "ERROR" string.
                # -The Null Trap: Fill the missing Active_Users with the mean value of that column.
                # -The Final Output: Print df.info() and the final cleaned DataFrame.
```

```python
df2 = pd.DataFrame({
    'Date': ['2023-01-01', '2023-01-02', 'Missing', '2023-01-04'],
    'Revenue': ['$1,200', '$2,500', 'ERROR', '$1,800'],
    'Active_Users': [150, np.nan, 200, 175]
})    
```

```python
df2
```

```python
df2.info()
```

```python
df2['Date'] = pd.to_datetime(df2['Date'], errors='coerce')
```

```python
df2.info() # datatype change of Date column
```

```python
df2['Revenue'] = df2['Revenue'].str.replace(r"[$,]", "", regex = True)
df2
```

```python
df2['Revenue'] = pd.to_numeric(df2['Revenue'], errors='coerce')
```

```python
df2.info()
```

```python
df2
```

```python
 df2["Active_Users"] = df2["Active_Users"].fillna(df2["Active_Users"].mean())
```

```python
1. Challenge 1: The "Sales Insights" Aggregator
-Goal: Master the groupby() and agg() functions.
-The Task: Provide this dataset of sales transactions. The intern must write a script that calculates two things:
-The total revenue per Category.
-The average quantity sold per Region.
Code:
# import pandas as pd

# sales_data = {
#     'Region': ['North', 'South', 'North', 'East', 'South', 'East'],
#     'Category': ['Electronics', 'Furniture', 'Electronics', 'Electronics', 'Furniture', 'Furniture'],
#     'Revenue': [1200, 800, 1500, 2100, 450, 700],
#     'Quantity': [3, 2, 4, 5, 1, 2]
# }
# df = pd.DataFrame(sales_data)
-Requirement: Output a summary DataFrame showing Total Revenue by Category.
```

```python
sales_data = {
    'Region': ['North', 'South', 'North', 'East', 'South', 'East'],
    'Category': ['Electronics', 'Furniture', 'Electronics', 'Electronics', 'Furniture', 'Furniture'],
    'Revenue': [1200, 800, 1500, 2100, 450, 700],
    'Quantity': [3, 2, 4, 5, 1, 2]
}
```

```python
sd = pd.DataFrame(sales_data)
```

```python
sd
```

```python
sd.groupby("Category")["Revenue"].sum()
```

```python
sd.groupby("Region")["Quantity"].mean()
```

```python
sd.groupby("Category").agg({
    "Revenue" : ["sum", "mean"],
})
```

```python
sd["Total_By_Category"] = sd.groupby("Category")["Revenue"].transform("sum")
```
